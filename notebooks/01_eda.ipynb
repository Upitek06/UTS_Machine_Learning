```python
import pandas as pd
import hashlib
from sklearn.model_selection import train_test_split

# 1. Load Data
df = pd.read_csv('../data/breast_cancer.csv')

# 2. Pembersihan Data (Drop kolom tidak berguna)
df = df.drop(['id', 'Unnamed: 32'], axis=1, errors='ignore')

# 3. Encoding Target (Malignant=1, Benign=0)
df['diagnosis'] = df['diagnosis'].map({'M': 1, 'B': 0})

# 4. Cek Ketidakseimbangan Kelas
print("Distribusi Target:")
print(df['diagnosis'].value_counts(normalize=True) * 100)

# 5. Isolasi Lingkungan (Split 80:20 Stratified)
X = df.drop('diagnosis', axis=1)
y = df['diagnosis']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)

# 6. Simpan Indeks Test Data (Bukti Kunci Data)
uji_path = '../data/split_uji.csv'
X_test.index.to_frame(name='index_test').to_csv(uji_path, index=False)

# Generate Hash untuk Laporan
with open(uji_path, "rb") as f:
    print("\nHash split_uji.csv (Masukkan ke README):", hashlib.md5(f.read()).hexdigest())
```
